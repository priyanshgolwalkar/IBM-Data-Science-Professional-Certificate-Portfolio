# Machine Learning with Python

This notebook demonstrates supervised machine learning using classification models.


```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
data = {
    "Study_Hours": [1,2,3,4,5,6,7,8],
    "Attendance": [60,65,70,75,80,85,90,95],
    "Pass": [0,0,0,1,1,1,1,1]
}

df = pd.DataFrame(data)
df
```

```python
X = df[["Study_Hours", "Attendance"]]
y = df["Pass"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
lr = LogisticRegression()
lr.fit(X_train, y_train)

y_pred_lr = lr.predict(X_test)
print("Logistic Regression Accuracy:", accuracy_score(y_test, y_pred_lr))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred_lr))
```

```python
rf = RandomForestClassifier()
rf.fit(X_train, y_train)

y_pred_rf = rf.predict(X_test)
print("Random Forest Accuracy:", accuracy_score(y_test, y_pred_rf))
```

## Model Comparison

- Logistic Regression and Random Forest were trained.
- Accuracy metrics were evaluated.
- Random Forest showed strong classification performance.

This demonstrates supervised classification workflow.
