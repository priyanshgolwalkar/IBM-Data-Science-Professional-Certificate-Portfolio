# Data Analysis with Python

This project demonstrates exploratory data analysis (EDA), data cleaning, and statistical insights using Pandas and Matplotlib.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

**Create Dataset with Missing Values**

```python
data = {
    "Age": [21, 22, np.nan, 24, 23],
    "Marks": [85, 90, 78, np.nan, 88],
    "Study_Hours": [2, 3, 4, 5, np.nan]
}

df = pd.DataFrame(data)
df
```

**Handle Missing Values**

```python
df.fillna(df.mean(numeric_only=True), inplace=True)
df
```

**Statistical Summary**

```python
df.describe()
```

```python
df.corr()
```
